# 特征预处理（归一化 + 标准化）

***

## 1. 基础概念
**概念与分类**
* 特征预处理：就是通过一些转换函数将数据转化成更加适合算法模型的特征数据过程
* 数值型数据的无量纲化
    * 归一化
    * 标准化
* 进行归一化/标准化的原因：
    * 特征的单位或大小差异较大，或者某特征的方差相比其他的特征要大出几个量级，容易影响（支配）目标结果，使得一些算法无法学习到其他特征
      
**特征预处理API**
* sklearn.preprocessing

## 2. 归一化
**概念与公式**
* 定义：通过对原始数据进行变换把数据映射到（默认为[0,1]）之间
* 公式：
  $$X' = \frac{x - min}{max - min} \quad   X'' = X' \times (mx - mi) + mi$$
  * *作用于每一列，其中$max$为一列的最大值，$min$为一列的最小值；*
  
  * *$X''$ 为最终结果，$mx$、$mi$分别为指定区间值（默认$mx$为1、$mi$为0）*

**归一化API**
* sklearn.preprocessing.MinMaxScarer(feature_range(0,1)...)
    * MinMaxScalar.fit_transform(X)
        * X: numpy array格式的数据[n_samples, n_features]
    * 返回值：转换后的形状相同的array   

**案例实现**
* 约会对象数据：
  
  相亲约会对象数据是一份男士数据，其具有三个特征：玩游戏所消耗时间的百分比、每年获得的飞行常客里程数和美洲小号的冰淇淋公升数。有一个所属类别，即被女士评价为：不喜欢didnt、魅力一般small和极具魅力large。从表格来看里程数、公升数和消耗时间比的差距很大（这样里程数对评价的影响会最大），但我们认为三个特征对其结算结果同等重要，因此我们需要对其进行归一化。(本数据放在"data/dating.txt"中)
  
| 里程数   | 公升数     | 消耗时间比 | 评价        |
|:----------:|:------------:|:------------:|:-------------:|
| 40920    | 8.326976   | 0.953952   | largeDoses  |
| 14488    | 7.153469   | 1.673904   | smallDoses  |
| 26052    | 1.441871   | 0.805124   | didntLike   |
| 75136    | 13.147394  | 0.428964   | didntLike   |
| 38344    | 1.669788   | 0.134296   | didntLike   |


* 代码思路
    * 实例化MinMaxScalar
    * 通过fit_transform转换 

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

def minmax_demo():
    """
    归一化演示
    :return: None
    """
    data = pd.read_csv("data/dating.txt")
    print(data)
    # 1. 实例化一个转换器类
    transfer = MinMaxScaler(feature_range=(2,3))
    # 2. 调用fit_transform
    X = data[['milage', 'Liters', 'Consumtime']]  
    X_scaled = transfer.fit_transform(X)
    print("最小值最大值归一化处理的结果：\n",X_scaled)

    return None
```

```python
minmax_demo()
```

**注意**
* 我们从公式角度可以关注到“归一化”主要是通过最大值和最小值的关系去进行计算的
* 那么如果数据中的异常点较多，可能会影响归一化的结果
    * 例如：其中有一个数据异常大，则会通过最大值 $max$ 影响归一化的结果

**总结**

* 综上所述，因为最大最小值容易受异常点的影响，所以这种方法的鲁棒性较差，只适合传统精确小数据的场景

# 3. 标准化

**概念与公式**
* 定义：通过对原始数据进行变换吧数据变化到均值为0，标准误差为1范围内
* 公式：
  $$X' = \frac{x - mean}{\sigma} \quad $$
  * 作用于每一列，mean为平均值，$\sigma$ 为标准化
* 面对异常点的优势
    * 对于对一化来说：如果出现了异常点，会影响其最大值和最小值 ，则结果显然会发生改变
    * 对于标准化来说：如果出现了异常点，由于其具有一定数据量，少量的异常点对平均值的影响并不大，从而方差改变较小

**标准化API**
* sklearn.preprocessing.StandardScaler()
    * 处理之后每列来说所有数据都聚集在均值0附近标准差为1
    * StandardScaler.fit_transform(X)
        * X：numpy array格式的数据[n_amples,n_features]
    * 返回值：转换后的形状相同的array 

**数据计算**
* 实例化StandardScaler
* 通过fit_transform转换

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

def stand_demo():
    """
    标准化演示
    :return: None
    """
    data = pd.read_csv("data/dating.txt")
    print(data)
    # 1.实例化一个转换器类
    transfer = StandardScaler()
    # 2. 调用fit_transform
    X = data[['milage','Liters','Consumtime']]
    X_Standard = transfer.fit_transform(X)
    print("标准化的结果：\n",X_Standard)
    print("每一列特征值的平均值：\n",transfer.mean_)     # 打印平均值笔记本
    print("每一列特征值的方差：\n",transfer.var_)     #打印方差笔记本
```

```python
stand_demo()
```

**总结**
* 综上所述，标准化的方法在已有样本足够多的情况下比较稳定，适合现代嘈杂大数据的场景
