# TF-IDF 文本特征提取

***

## 1. 基础概念
* TF-IDF的主要思想是：如果某个词或短语在一篇文章中的出现概率高，并且在其他文章中很少出现，则认为此词语或者短语具有很良好的类别区分能力，适合用于分类
* TF-IDF的作用： 用以评估一字词对于一个文件集或一个语料库中的其中一份文件的重要程度。

## 2. 公式
TF-IDF公式为（最终得出的结果可以理解为重要程度）：
 $$tfidf_{i,j} = tf_{i,j} \times idf_i$$
其中：
* 词频（term frequency, tf）：指的是某一个给定的词语在该文件中出现的频率
* 逆向文档频率（inverse document frequency, idf）：是指一个词语普遍重要性的度量。
    * 某一特定词语的idf，可以由总文件数目除以包含词语之文件的数目，再将得到的商取以10为底的对数得到

## 3. 应用代码

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import jieba

def cut_word(text):
    """
    对中文进行分词
    "我爱北京天安门" --------> "我 爱 北京 天安门"
    :param text: 待分词的原始中文字符串，例如 "我爱北京天安门"
    :return:分词后用单个空格隔开的结果字符串，例如 "我 爱 北京 天安门"
    """
    # 用jieba对中文字符串进行分词
    text = " ".join(jieba.cut(text))   # join可以直接可以直接接收生成器

    return text

def text_chinese_tfidf_demo():
    """
    对中文进行特征抽取
    :return: None
    """
    data = ["今天很残酷，明天更残酷，后天很美好，但绝大部分是死在明天晚上，所以每个人不要放弃今天。",
            "我们看到的从很远星系来的光是几百万年之前发出的，这样当我们看到宇宙时，我们是在看它的过去。",
            "如果只用一种方式了解某样事物，你就不会真正了解它。了解事物真正含义的秘密取决于如何将其与我们所了解的事物想联系。"]
    # 将原始数据转换成分好词的形式
    text_list = []
    for sent in data:
        text_list.append(cut_word(sent))
    print(text_list)

    # 1.实例化一个转换器类
    transfer = TfidfVectorizer(stop_words=['一种','不会','不要'])    #停用词表
    # 2.调用fit_transform
    data_matrix = transfer.fit_transform(text_list)
    print("文本特征抽取的结果：\n", data_matrix.toarray())
    print("返回特征名字：\n", transfer.get_feature_names_out())

    return None
```

```python
text_chinese_tfidf_demo()
```
