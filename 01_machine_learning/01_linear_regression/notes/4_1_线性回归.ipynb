# 线性回归

***

## 一、线性回归的基本概念

### 1.1 线性回归概述

**定义**：线性回归用于建立特征（自变量）与目标值（因变量）之间的线性关系模型。通过数据拟合出一条直线（或超平面）来预测连续值。

### 1.2 应用实例

**数据**：工资和年龄（2个特征）  
**目标**：预测银行会贷款给客户多少钱（标签）  
**考虑问题**：工资和年龄都会影响最终银行贷款的结果，它们各自有多大的影响？（即参数估计）

**示例数据**：
| 工资 | 年龄 | 额度 |
|------|------|------|
| 4000 | 25   | 20000 |
| 8000 | 30   | 70000 |
| 5000 | 28   | 35000 |
| 7500 | 33   | 50000 |
| 12000| 40   | 85000 |

**通俗解释**：
- $X_1, X_2$：两个特征（年龄、工资）
- $Y$：标签（银行最终会借给我们多少钱）

## 二、数学建模

### 2.1 线性回归模型
假设$θ_1$是年龄的参数，$θ_2$是工资的参数，拟合的平面为：
$$ h_θ(x) = θ_0 + θ_1x_1 + θ_2x_2 $$
其中$θ_0$是偏置项（截距）。

**整合形式**（对于n个特征）：
$$ h_θ(x) = θ_0 + \sum_{i=1}^n θ_ix_i = θ^Tx $$

### 2.2 误差分析
真实值$y^{(i)}$和预测值$h_θ(x^{(i)})$之间存在差异，用$ε^{(i)}$表示误差。

对于每个样本：
$$ y^{(i)} = θ^Tx^{(i)} + ε^{(i)} $$

**误差假设**（高斯分布假设）：
- 误差$ε^{(i)}$是独立且同分布的
- 服从均值为0，方差为$σ^2$的高斯分布

**假设的通俗解释**：
1. **独立**：张三和李四一起来贷款，他俩没关系
2. **同分布**：他俩都来得是我们假定的这家银行
3. **高斯分布**：银行可能会多给，也可能会少给，但是绝大多数情况下这个浮动不会太大，极小情况下浮动会比较大，符合正常情况

## 三、极大似然估计推导

### 3.1 从误差分布到似然函数
由于误差服从高斯分布：
$$ ε^{(i)} ∼ N(0, σ^2) $$
即：
$$ p(ε^{(i)}) = \frac{1}{\sqrt{2π}σ} \exp\left(-\frac{(ε^{(i)})^2}{2σ^2}\right) $$

代入$ε^{(i)} = y^{(i)} - θ^Tx^{(i)}$：
$$ p(y^{(i)}|x^{(i)}; θ) = \frac{1}{\sqrt{2π}σ} \exp\left(-\frac{(y^{(i)} - θ^Tx^{(i)})^2}{2σ^2}\right) $$

### 3.2 似然函数
对于m个样本：
$$ L(θ) = \prod_{i=1}^m p(y^{(i)}|x^{(i)}; θ) $$
解释：什么样的参数跟我们的数据组合后恰好是真实值。

### 3.3 对数似然函数
$$ l(θ) = \log L(θ) = \sum_{i=1}^m \log p(y^{(i)}|x^{(i)}; θ) $$
解释：乘法难解，加法就容易了，对数里面乘法可以转换成加法。

展开化简：
$$ l(θ) = m\log\frac{1}{\sqrt{2π}σ} - \frac{1}{2σ^2} \sum_{i=1}^m (y^{(i)} - θ^Tx^{(i)})^2 $$
解释：希望$l(θ)$越大越好
## 四、目标函数与最小二乘法

### 4.1 目标函数
要使似然函数最大，相当于使负对数似然最小。去掉常数项，得到目标函数：
$$ J(θ) = \frac{1}{2} \sum_{i=1}^m (y^{(i)} - θ^Tx^{(i)})^2 $$
这就是**最小二乘法**的损失函数。

有时候为了方便梯度计算，会使用均方误差形式：
$$ J(θ) = \frac{1}{2m} \sum_{i=1}^m (y^{(i)} - h_θ(x^{(i)}))^2 $$

### 4.2 直接求解（正规方程）
对于线性回归，可以直接通过求偏导得到解析解：
$$ \frac{\partial J(θ)}{\partial θ} = 0 $$

求解得到：
$$ θ = (X^TX)^{-1}X^TY $$

其中$X$是设计矩阵（包含所有样本特征），$Y$是目标值向量。

## 五、模型评估

### 5.1 评估指标
最常用的评估指标是**决定系数 $R^2$**：
$$ R^2 = 1 - \frac{SS_{res}}{SS_{tot}} $$

其中：
- $SS_{res} = \sum_{i=1}^m (y^{(i)} - \hat{y}^{(i)})^2$（残差平方和）
- $SS_{tot} = \sum_{i=1}^m (y^{(i)} - \bar{y})^2$（总平方和）

**解释**：
- $R^2$的取值范围越接近于1，表示模型拟合得越好
- $SS_{res}$类似方差项，表示模型未能解释的方差

## 六、梯度下降优化

### 6.1 梯度下降的基本思想

当我们得到一个目标函数后，要如何进行求解？一定是可以直接求解的吗？并不是，线性回归的可解是一个特例。
因此，当目标函数无法直接求解时（或直接求解效率低时），我们需要使用梯度下降迭代优化。

**梯度下降的直观比喻**：
- 寻找山谷的最低点，也就是我们的目标函数终点
- 下山分几步走：
  1. **找方向**：找到当前最合适的方向（梯度方向）
  2. **定步长**：走一小步，走快了该"跌倒"了（学习率控制）
  3. **更新参数**：按照方向与步伐更新参数

### 6.2 梯度计算
目标函数$J(θ_0,θ_1) = \frac{1}{2m} \sum_{i=1}^m (y^{(i)} - h_θ(x^{(i)}))^2$

梯度计算公式：
$$ \frac{\partial J(θ)}{\partial θ_j} = \frac{1}{m} \sum_{i=1}^m (h_θ(x^{(i)}) - y^{(i)})x_j^{(i)} $$

### 6.3 三种梯度下降算法

#### 6.3.1 批量梯度下降（Batch Gradient Descent, BGD）
每次迭代使用所有m个样本计算梯度：
$$ θ_j := θ_j - α \frac{1}{m} \sum_{i=1}^m (h_θ(x^{(i)}) - y^{(i)})x_j^{(i)} $$

**特点**：
- 优点：容易得到最优解
- 缺点：每次考虑所有样本，速度很慢

#### 6.3.2 随机梯度下降（Stochastic Gradient Descent, SGD）
每次迭代只随机选择一个样本计算梯度：
$$ θ_j := θ_j - α (h_θ(x^{(i)}) - y^{(i)})x_j^{(i)} $$

**特点**：
- 优点：每次找一个样本，迭代速度快
- 缺点：不一定每次都朝着收敛的方向

#### 6.3.3 小批量梯度下降（Mini-batch Gradient Descent, MBGD）
每次迭代使用一小批（batch）样本计算梯度：
$$ θ_j := θ_j - α \frac{1}{|B|} \sum_{i∈B} (h_θ(x^{(i)}) - y^{(i)})x_j^{(i)} $$

**特点**：
- 实际应用中最常用
- 结合了BGD的稳定性和SGD的速度

### 6.4 超参数选择

#### 6.4.1 学习率（α）
- **作用**：控制参数更新的步长
- **影响**：对结果会产生巨大的影响
- **选择策略**：
  - 一般设置较小值
  - 可以尝试从较小的值开始，不行再减小
  - 常见范围：0.001、0.01、0.1等

#### 6.4.2 批处理大小（batch size）
- **常用值**：32、64、128等
- **考虑因素**：
  - 内存限制
  - 计算效率
  - 收敛稳定性

## 七、线性回归的优缺点

### 7.1 优点
1. **简单易懂**：模型形式简单，解释性强
2. **计算效率高**：无论是解析解还是梯度下降都比较高效
3. **理论基础扎实**：有严格的数学推导和统计理论基础
4. **可解释性强**：每个特征的系数直接反映了其对目标的影响程度
5. **基准模型**：常作为其他复杂模型的性能基准

### 7.2 缺点
1. **线性假设限制**：只能建模线性关系，无法处理非线性关系
2. **对异常值敏感**：最小二乘法对异常值非常敏感
3. **多重共线性问题**：当特征间高度相关时，可能导致系数估计不稳定
4. **假设条件严格**：需要满足高斯马尔可夫假设（线性、独立同分布、同方差等）

## 八、实际应用建议

### 8.1 数据预处理
1. **特征缩放**：使用标准化或归一化处理特征
2. **特征工程**：可以构造多项式特征扩展线性模型的能力
3. **异常值处理**：检测和处理异常值，提高模型鲁棒性

### 8.2 正则化（补充内容）
为了防止过拟合，可以对线性回归添加正则化项：

1. **岭回归（Ridge Regression, L2正则化）**：
   $$ J(θ) = \frac{1}{2m} \sum_{i=1}^m (y^{(i)} - h_θ(x^{(i)}))^2 + λ\sum_{j=1}^n θ_j^2 $$

2. **LASSO回归（L1正则化）**：
   $$ J(θ) = \frac{1}{2m} \sum_{i=1}^m (y^{(i)} - h_θ(x^{(i)})^2 + λ\sum_{j=1}^n |θ_j| $$

### 8.3 模型验证
1. **交叉验证**：使用k折交叉验证选择最优超参数
2. **误差分析**：分析训练误差和验证误差的差异
3. **残差分析**：检查残差是否符合高斯分布假设

## 九、总结

线性回归作为机器学习中最基础、最重要的算法之一，具有以下核心要点：

1. **模型本质**：通过最小化预测值与真实值之间的平方误差，找到最佳的线性拟合
2. **数学基础**：基于高斯分布假设，通过极大似然估计推导出最小二乘法
3. **优化方法**：可以通过解析解（正规方程）或梯度下降进行求解
4. **评估指标**：$R^2$是衡量模型拟合优度的核心指标
5. **实用技巧**：根据数据规模选择合适的梯度下降算法，调整学习率和批大小

**机器学习实践建议**
> 常规套路：我交给机器一堆数据，然后告诉它什么样的学习方式是对的（目标函数），然后让它朝着这个方向去做。一口吃不成个胖子，我们要静悄悄的一步步的完成迭代（每次优化一点点，累积起来就是个大成绩了）。
