
**多分类需要转换为多个二分类**

* 代码片段如下：

```Python
for label_index, unique_label enumerate(self.unique_labels):
            current_initial_theta = np.copy(self.theta[label_index].reshape(num_features, 1))
            current_labels = (self.labels == unique_label).astype(float)
            (current_theta, cost_history) = LogisticRegression.gradient_desent(self.data, current_labels, 
                                                                               current_initial_theta, max_iterations)
```
* 以三分类为例，这一段循环可以理解为下：
```
外层循环 3 次：
│
├── 第 1 次：unique_label = 'A', label_index = 0
│   ├── 从 θ[0] 取出一行 → reshape 成列向量 (n, 1)
│   ├── 标签转换：y == 'A' → True→1, False→0
│   └── 二分类训练 → 拿出训练迭代后的theta和cost（更新 θ[0] ）
│ 
├── 第 2 次：unique_label = 'B', label_index = 1
│   ├── 从 θ[1] 取出一行 → 列向量 (n, 1)
│   ├── 标签转换：y == 'B' → True→1, False→0
│   └── 二分类训练 → 拿出训练迭代后的theta和cost（更新 θ[1]）
│
└── 第 3 次：unique_label = 'C', label_index = 2
    ├── 从 θ[2] 取出一行 → 列向量 (n, 1)
    ├── 标签转换：y == 'C' → True→1, False→0
    └── 二分类训练 → 拿出训练迭代后的theta和cost（更新 θ[2]）
```
